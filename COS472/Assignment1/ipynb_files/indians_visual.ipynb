```python
# Colby Wirth
# COS 472 Assignment 1 
# September 24 2024

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

cleaned_indians= pd.read_csv("../csv_files/indians_cleaned.csv")

cleaned_indians.head()
```

```python
plt.plot(cleaned_indians['BMI'], cleaned_indians['BloodPressure'], 'bo')


z = np.polyfit(cleaned_indians['BMI'], cleaned_indians['BloodPressure'], 1)

p = np.poly1d(z)

plt.scatter('BMI', 'BloodPressure')
plt.plot(cleaned_indians['BMI'], p(cleaned_indians['BMI']))

plt.title("BMI vs Blood Pressure of the Native American Group")
plt.xlabel("BMI")
plt.ylabel("Blood Pressure")

plt.show()
```

```python
z = np.polyfit(cleaned_indians['Age'], cleaned_indians['BloodPressure'], 1)
p = np.poly1d(z)

plt.scatter(cleaned_indians['Age'], cleaned_indians['BloodPressure'], c=cleaned_indians['BloodPressure'], cmap=plt.cm.plasma)

plt.plot(cleaned_indians['Age'], p(cleaned_indians['Age']), color='black')

plt.title("Age vs Blood Pressure of the Native American Group")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.show()
```

```python
sns.histplot(data=cleaned_indians, x='Pregnancies')
plt.title("Pregnancy Distribution")
plt.xlabel("Number Pregnancies")
plt.ylabel("Number of people with Pregnancies")
```

```python
bins = [20, 30, 40, 50, 60, 70, 80, 90]

labels = ['21-30', '31-40', '41-50', '51-60', '61-70', '71-80', '81+']

cleaned_indians['age_buckets'] = pd.cut(cleaned_indians['Age'], bins=bins, labels=labels, right=False)

plt.xlabel('Ages')

sns.boxplot(data=cleaned_indians, x='age_buckets', y='Pregnancies')
```

```python
plt.scatter(cleaned_indians['Insulin'], cleaned_indians['Glucose'], color='red')

plt.title("Insulin vs Glucose of the Pima Native American Group")
plt.xlabel("Insulin")
plt.ylabel("Glucose")

plt.show()
```
