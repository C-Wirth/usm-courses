```python
indian_df = pd.read_csv("../csv_files/indians.csv")

indian_df.shape
```

```python
#All weight values are null drop values
indian_df.drop('Weight', axis=1, inplace=True)
```

```python
#All values = '-1', provide no useful data for analysis
indian_df.drop('Occuption', axis=1, inplace=True)
```

```python
indian_df.info()
```

```python
indian_df.duplicated().sum()
```

```python
indian_df = indian_df.drop_duplicates()
```

```python
indian_df.isnull().sum()
```

```python
# indian_df_cleaned = indian_df.dropna()

indian_df['Glucose'] = indian_df['Glucose'].fillna(indian_df['Glucose'].mean())

indian_df['BloodPressure'] = indian_df['BloodPressure'].fillna(indian_df['BloodPressure'].mean())

indian_df['Insulin'] = indian_df['Insulin'].fillna(indian_df['Insulin'].mean())

indian_df['BMI'] = indian_df['BMI'].fillna(indian_df['BMI'].mean())

indian_df['DiabetesPedigreeFunction'] = indian_df['DiabetesPedigreeFunction'].fillna(indian_df['DiabetesPedigreeFunction'].mean())
```

```python
indian_df.isnull().sum()
```

```python
indian_df.to_csv('../csv_files/indians_cleaned.csv', index=False)
```
