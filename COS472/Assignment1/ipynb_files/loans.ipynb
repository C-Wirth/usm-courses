```python
#Create a dataframe by parsing loan.csv with the pd.read_csv function

#The pd.read_csv function  parses comma separated values into columns, and each line into a row.

loan_df = pd.read_csv("loan.csv")
```

```python
#The describe() function gives useful statistical data regarding each attribute in the columns.

#The 'T' transposes the dataframe as you would a matrix.

loan_df.describe().T
```

```python
#The drop function removes the indicated columns,
#  axis 1 refers to columns (axis 0 would refer to rows)
# 'inplace' refers to chaning the df referenced in memory - false would return a new df

#We are dropping the duplicate attribute, and the empty atrribute present in the df specifically here

loan_df.drop(['days.with.cr.line.1', 'history'], axis=1, inplace=True)
```

```python
#Verify the change have beenn made
loan_df.info()
```

```python
#Now we take count of how many dupliacted rows exist in the df

loan_df.duplicated().sum()
```

```python
#remove the duplicated rows

loan_df = loan_df.drop_duplicates()
```

```python
#Verify the 13 duplicated rows have been dropped
loan_df.shape
```

```python
#Now we must handle null values

loan_df.isnull().sum()
```

```python
#Identify nominal values for the purpose attribute
loan_df['purpose'].value_counts()
```

```python
#Use a standard value to fill nan values for purpose attriburte
loan_df['purpose'].fillna('debt_consolidation',inplace=True)
```

```python
#verify that the nan values for purpose have been filled
loan_df.isnull().sum()
```

```python
#remvoe all rows with nan values
df_cleaned = loan_df.dropna()
```

```python
#verify no nan values remain
df_cleaned.isnull().sum()
```

```python
#Verify info
df_cleaned.shape
```

```python
#View info of cleaned df
df_cleaned.info()
```

```python
#Now we must change the object data listed under purpose into 1s and 0s, represented Boolean values

# List of columns to fill missing values with their mean
columns = ['credit.policy', 'int.rate', 'installment', 'log.annual.inc', 
           'dti', 'fico', 'days.with.cr.line', 'revol.bal', 'revol.util', 
           'inq.last.6mths', 'delinq.2yrs', 'pub.rec', 'not.fully.paid']

# Iterate over the columns and fill missing values with the mean
for col in columns:
    loan_df[col] = loan_df[col].fillna(loan_df[col].mean())
```

```python
#Create a temporary df to house the new attributes

df_enconde = pd.get_dummies(loan_df, columns=['purpose'], dtype=int)
```

```python
# adjust order, put the label column 'not fully paid' to last

#pop target column
col= df_enconde.pop('not.fully.paid')

#add it back to df
df_enconde['not.fully.paid'] = col
```
